```python
# Import libraries and set up paths
import os
from pathlib import Path
import pandas as pd
import numpy as np
from pprint import pprint

ROOT = Path('.')
DATA_DIR = ROOT / '../../MachineLearningCVE'
OUTPUT_DIR = ROOT / '../../MachineLearningCVE_Cleaned'
OUTPUT_DIR.mkdir(exist_ok=True)

# discover CSV files
csv_files = sorted([p for p in DATA_DIR.iterdir() if p.suffix.lower() == '.csv'])
print(f'Found {len(csv_files)} CSV files in', DATA_DIR)
pprint([p.name for p in csv_files])

# load into DataFrames dictionary (do not modify originals on disk)
raw_dfs = {}
for p in csv_files:
    try:
        df = pd.read_csv(p)
    except Exception as e:
        print(f'Failed to read {p.name}:', e)
        continue
    raw_dfs[p.name] = df

# show initial shapes and headers
for name, df in raw_dfs.items():
    print(f'-- {name} : shape={df.shape}')
    print('  columns=', list(df.columns)[:10])
```

```python
# Remove duplicate network records or repeated entries
# 1) Remove duplicate rows (exact duplicates)
cleaned_dfs = {}  # will hold cleaned DataFrames
duplicates_info = {}
for name, df in raw_dfs.items():
    before = len(df)
    # keep first occurrence, drop exact duplicate rows
    deduped = df.drop_duplicates(keep='first').reset_index(drop=True)
    after = len(deduped)
    duplicates_info[name] = {'before': before, 'after': after, 'removed': before - after}
    cleaned_dfs[name] = deduped

# print summary and show sample duplicates for each file (if any)
for name, info in duplicates_info.items():
    print(f"{name}: rows before={info['before']}, after={info['after']}, removed={info['removed']}")
    if info['removed'] > 0:
        print(' Sample duplicate rows (first 3 after dedup detection):')
        # show rows that were duplicates by comparing with original index
        df_orig = raw_dfs[name]
        # find duplicates boolean mask
        dup_mask = df_orig.duplicated(keep='first')
        display(df_orig[dup_mask].head(3))

# Keep cleaned_dfs for further steps
len(cleaned_dfs)
```

```python
# 2) Handle missing or corrupted values
from ipaddress import ip_address

def is_valid_ip(x):
    if pd.isna(x):
        return False
    try:
        ip_address(str(x))
        return True
    except Exception:
        return False

for name, df in list(cleaned_dfs.items()):
    print(f'Processing missing values for {name} - initial shape {df.shape}')
    missing_counts = df.isna().sum()
    print('  columns with missing values (top 10):')
    print(missing_counts[missing_counts>0].sort_values(ascending=False).head(10))

    # conservative rules: drop rows missing essential network fields if present
    essential_cols = []
    for c in ['Source IP','Destination IP','Source Port','Destination Port','Timestamp','timestamp','ts']:
        if c in df.columns:
            essential_cols.append(c)
    if essential_cols:
        before = len(df)
        df = df.dropna(subset=essential_cols).reset_index(drop=True)
        after = len(df)
        print(f'  dropped {before-after} rows missing essential columns: {essential_cols}')

    # attempt light normalization for IP columns: mark invalid IPs as NaN
    for col in df.columns:
        if 'ip' in col.lower() or 'addr' in col.lower():
            # vectorized validation might be slow; do small sample then fallback to apply
            df[col] = df[col].where(df[col].apply(is_valid_ip), other=np.nan)

    cleaned_dfs[name] = df

len(cleaned_dfs)

# end missing handling
```

```python
# 3) Convert data types (numeric, categorical, datetime)
import dateutil

def try_parse_datetime(s):
    try:
        return pd.to_datetime(s, errors='coerce', utc=True)
    except Exception:
        return pd.NaT

for name, df in list(cleaned_dfs.items()):
    print('Converting types for', name)
    # datetime columns detection
    for col in df.columns:
        if 'time' in col.lower() or 'date' in col.lower() or 'ts' == col.lower():
            df[col] = try_parse_datetime(df[col])
    # numeric conversion for obvious numeric columns
    for col in df.columns:
        if any(k in col.lower() for k in ['size','bytes','length','count','port','ttl','duration']):
            df[col] = pd.to_numeric(df[col], errors='coerce')
    cleaned_dfs[name] = df
    print('  after conversion shape', df.shape)

'done'
```

```python
# Normalize IPs, ports, and protocol formats for consistency
# 4) Normalize IPs, ports, and protocol formats for consistency
import re
from ipaddress import ip_address, IPv4Address, IPv6Address

def normalize_ip(x):
    if pd.isna(x):
        return x
    s = str(x).strip()
    # remove surrounding brackets for IPv6 [::1]:80 -> ::1:80
    s = s.strip('[]')
    # if ip:port, split off port
    if ':' in s and s.count(':') == 1 and re.match(r'^[0-9]+$', s.split(':')[-1]):
        s = s.split(':')[0]
    # for IPv6 with port like '::1:80' we cannot reliably split, so try ip_address validation
    try:
        # this will succeed for plain IPs
        ip_address(s)
        return s
    except Exception:
        # fallback: try to extract first token that looks like an IP
        tokens = re.split('[,;\/]+', s)
        for t in tokens:
            try:
                ip_address(t)
                return t
            except Exception:
                continue
        return np.nan

def normalize_port(x):
    if pd.isna(x):
        return x
    try:
        s = str(x).strip()
        # sometimes ports come as '80/HTTP' or '80.0'
        s = re.split('[^0-9]', s)[0]
        p = int(s)
        if 0 <= p <= 65535:
            return p
        return np.nan
    except Exception:
        return np.nan

def normalize_proto(x):
    if pd.isna(x):
        return x
    s = str(x).strip().lower()
    # common replacements
    s = s.replace('ip','ip').replace('tcp','tcp').replace('udp','udp')
    # if numeric proto (e.g., '6' for TCP), keep as string of number
    if s.isdigit():
        return s
    return s

# apply normalization heuristics to detected columns
for name, df in list(cleaned_dfs.items()):
    print('Normalizing formats for', name)
    # IP-like columns
    for col in [c for c in df.columns if 'ip' in c.lower() or 'addr' in c.lower()]:
        before = df[col].nunique(dropna=True) if col in df.columns else 0
        df[col] = df[col].apply(normalize_ip) if col in df.columns else df.get(col)
        after = df[col].nunique(dropna=True) if col in df.columns else 0
        print(f'  {col}: uniques before={before} after={after}')
    # Port-like columns
    for col in [c for c in df.columns if 'port' in c.lower()]:
        df[col] = df[col].apply(normalize_port) if col in df.columns else df.get(col)
        print(f'  normalized ports in {col} (non-numeric -> NaN)')
    # Protocol columns
    for col in [c for c in df.columns if 'proto' in c.lower() or 'protocol' in c.lower()]:
        df[col] = df[col].apply(normalize_proto) if col in df.columns else df.get(col)
        print(f'  normalized protocol values in {col} (sample=', df[col].dropna().unique()[:5], ')')
    cleaned_dfs[name] = df

'normalized'
```

```python
# 5) Remove outliers in numeric columns conservatively
for name, df in list(cleaned_dfs.items()):
    print('Outlier removal for', name)
    for col in df.select_dtypes(include=[np.number]).columns:
        # skip small integer id-like columns by name heuristics
        if any(k in col.lower() for k in ['id','index','num']) and df[col].nunique() > 1000:
            continue
        # use IQR method
        series = df[col].dropna()
        if len(series) < 10:
            continue
        q1 = series.quantile(0.25)
        q3 = series.quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 3 * iqr
        upper = q3 + 3 * iqr
        before = len(df)
        df = df[(df[col].isna()) | ((df[col] >= lower) & (df[col] <= upper))]
        after = len(df)
        if before-after > 0:
            print(f'  removed {before-after} rows as outliers from {col}')
    cleaned_dfs[name] = df.reset_index(drop=True)

'outliers removed'
```

```python
# 6) Aggregate flows by common keys (conservative - create aggregated copies)
aggregated_dfs = {}
for name, df in cleaned_dfs.items():
    # choose grouping keys if available
    keys = []
    for k in ['Source IP','SourceIP','src_ip','src','source','saddr']:
        if k in df.columns:
            keys.append(k)
            break
    for k in ['Destination IP','DestinationIP','dst_ip','dst','destination','daddr']:
        if k in df.columns:
            keys.append(k)
            break
    for k in ['protocol','proto','Protocol']:
        if k in df.columns:
            keys.append(k)
            break
    if len(keys) >= 2:
        # aggregate numeric columns by sum/mean where appropriate
        agg_num = {c: 'sum' for c in df.select_dtypes(include=[np.number]).columns}
        try:
            agg_df = df.groupby(keys, dropna=False).agg(agg_num).reset_index()
            aggregated_dfs[name] = agg_df
            print(f'  aggregated {name} -> {agg_df.shape} using keys {keys}')
        except Exception as e:
            print('  aggregation failed for', name, e)
    else:
        print('  not enough keys to aggregate for', name, 'keeping original')
        aggregated_dfs[name] = df.copy()

'aggregated'
```

```python
# 7) Drop columns that leak labels or are not useful for prediction
# Conservative default list - will drop columns commonly used as unique IDs or file-specific meta
drop_like = ['Flow ID','id','Id','index','timestamp_ms','frame_number']
clean_for_model = {}
for name, df in aggregated_dfs.items():
    to_drop = [c for c in df.columns if any(d.lower() == c.lower() for d in drop_like)]
    if to_drop:
        print(f'{name}: dropping columns {to_drop}')
    model_df = df.drop(columns=to_drop, errors='ignore').copy()
    clean_for_model[name] = model_df.reset_index(drop=True)

'dropped'
```

```python
# 8) Encode categorical variables (protocols, flags) into numeric form
from sklearn.preprocessing import LabelEncoder
encoded_for_model = {}
label_encoders = {}
for name, df in clean_for_model.items():
    df2 = df.copy()
    for col in df2.select_dtypes(include=['object', 'category']).columns:
        if df2[col].nunique() <= 200:
            le = LabelEncoder()
            try:
                df2[col] = le.fit_transform(df2[col].astype(str))
                label_encoders[(name, col)] = le
            except Exception as e:
                print('  encoding failed for', name, col, e)
    encoded_for_model[name] = df2

'encoded'
```

```python
# 9) Scale numeric features using RobustScaler (handles outliers better)
from sklearn.preprocessing import RobustScaler
scaled_for_model = {}
scalers = {}
for name, df in encoded_for_model.items():
    df3 = df.copy()
    num_cols = df3.select_dtypes(include=[np.number]).columns.tolist()
    if num_cols:
        # Prepare numeric matrix; fill NA for transformer but check for available samples
        X = df3[num_cols].fillna(0)
        # If there are zero rows or all values are NaN (after fillna -> check original), skip scaling
        if X.shape[0] == 0 or df3[num_cols].dropna(how='all').shape[0] == 0:
            print(f"Skipping scaling for {name}: no numeric samples available (shape={X.shape})")
            scalers[name] = None
        else:
            try:
                scaler = RobustScaler()
                df3[num_cols] = scaler.fit_transform(X)
                scalers[name] = (scaler, num_cols)
            except Exception as e:
                print(f"Scaling failed for {name}: {e}")
                scalers[name] = None
    else:
        print(f"No numeric columns to scale for {name}")
        scalers[name] = None
    scaled_for_model[name] = df3

"scaled"
```

```python
# 10) Validate logical consistency and basic labeling/balancing
from collections import Counter
labeled = {}
for name, df in scaled_for_model.items():
    print('Validating', name)
    # basic checks: no negative durations
    for col in df.select_dtypes(include=[np.number]).columns:
        if any('duration' in c.lower() for c in [col]):
            neg = (df[col] < 0).sum()
            if neg>0:
                print(f'  found {neg} negative values in {col}, setting to abs()')
                df[col] = df[col].abs()
    # labeling: if 'Label' or 'label' or 'attack' in columns, normalize to binary malicious/benign
    label_col = None
    for c in df.columns:
        if c.lower() in ['label','labels','attack','class']:
            label_col = c
            break
    if label_col is not None:
        # create binary label: 1 for malicious-like strings
        vals = df[label_col].astype(str).str.lower()
        malicious = vals.str.contains('attack|malicious|ddos|portscan|infiltr|webatt')
        df['__label__'] = malicious.astype(int)
        print('  label distribution:', Counter(df['__label__']))
        labeled[name] = df
    else:
        labeled[name] = df

'validated'
```

```python
# 11) Save cleaned DataFrames to OUTPUT_DIR and print old vs new shapes/headers
# Ensure OUTPUT_DIR exists
try:
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
except Exception as e:
    raise RuntimeError(f"Failed to create OUTPUT_DIR {OUTPUT_DIR}: {e}")

to_save = {}
# prioritize labeled -> scaled datasets if present
source_map = labeled if labeled else scaled_for_model
for name, df in source_map.items():
    out_name = name.replace('.csv', '_cleaned.csv')
    out_path = OUTPUT_DIR / out_name
    try:
        # For safety, write with index=False
        df.to_csv(out_path, index=False)
        to_save[name] = {'path': str(out_path), 'shape': df.shape, 'columns': list(df.columns)[:20]}
        print(f'Saved {out_path.name}: shape={df.shape}')
    except Exception as e:
        print(f'Failed to save {out_path}: {e}')

# print comparison with raw files when possible
for orig_name, info in to_save.items():
    raw_shape = raw_dfs.get(orig_name).shape if orig_name in raw_dfs else None
    print(orig_name, ' raw_shape=', raw_shape, ' cleaned_shape=', info['shape'])

"saved"
```
