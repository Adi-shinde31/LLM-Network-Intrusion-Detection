```python
# Path to your MachineLearningCSV folder
DATA_PATH = "./MachineLearningCVE/"

# Correct chronological order of files
file_order = [
    "Monday-WorkingHours.pcap_ISCX.csv",
    "Tuesday-WorkingHours.pcap_ISCX.csv",
    "Wednesday-workingHours.pcap_ISCX.csv",
    "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv",
    "Friday-WorkingHours-Morning.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-DDoS.pcap_ISCX.csv"
]
```

```python
overall_summary = []
```

```python
for file in file_order:
    filepath = os.path.join(DATA_PATH, file)
    print("="*70)
    print(f"📂 Processing file: {file}")

    # Load CSV
    df = pd.read_csv(filepath)

    # Clean column names
    df.columns = df.columns.str.strip()

    # Replace inf with NaN, drop NaNs
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.dropna()

    # Summary
    print("Shape:", df.shape)

    # Data types
    print("\nData types:\n", df.dtypes.head(10))

    # Summary of numeric features
    print("\nSummary stats:\n", df.describe().T.head(10))

    if "Label" in df.columns:
        label_counts = df["Label"].value_counts()
        benign_count = label_counts.get("BENIGN", 0)
        attack_count = df.shape[0] - benign_count

        print("\nTop attack types:")
        print(label_counts.head(10))
        print(f"\nBenign: {benign_count}, Attacks: {attack_count}")

        # Store for summary
        overall_summary.append({
            "file": file,
            "rows": df.shape[0],
            "benign": benign_count,
            "attacks": attack_count,
            "attack_types": len(label_counts) - (1 if "BENIGN" in label_counts else 0)
        })

        # Plot attack distribution for this day
        plt.figure(figsize=(10,5))
        label_counts.plot(kind="bar")
        plt.title(f"Attack Distribution - {file}")
        plt.ylabel("Count")
        plt.tight_layout()
        plt.show()
    else:
        print("⚠️ No 'Label' column found!")
```

```python
print("="*70)
print("\n📊 Overall Summary")
summary_df = pd.DataFrame(overall_summary)
print(summary_df)
```

```python
print("\nGrand totals:")
print("Total rows:", summary_df['rows'].sum())
print("Total benign:", summary_df['benign'].sum())
print("Total attacks:", summary_df['attacks'].sum())
```

```python
# 📊 Overall benign vs attack visualization
plt.figure(figsize=(6,4))
summary_df[["benign", "attacks"]].sum().plot(kind="bar")
plt.title("Overall Benign vs Attacks")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
