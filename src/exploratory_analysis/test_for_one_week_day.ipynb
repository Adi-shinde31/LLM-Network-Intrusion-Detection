```python
import pandas as pd
```

```python
file = "MachineLearningCVE/Monday-WorkingHours.pcap_ISCX.csv"
df = pd.read_csv(file)
```

```python
# Basic dataset info
print("Shape:", df.shape)
print("\nColumns:\n", df.columns.tolist()[:20], "...")  # first 20 column names
print("\nFirst rows:\n", df.head())
```

```python
# Check missing values
print("\nMissing values:\n", df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
# Check unique labels
df.columns = df.columns.str.strip()
print("\nLabel distribution:\n", df['Label'].value_counts())
```

```python
import matplotlib.pyplot as plt
```

```python
df['Label'].value_counts().plot(kind='bar', figsize=(12,6))
plt.title("Traffic Label Distribution (Monday)")
plt.ylabel("Count")
plt.show()
```
