```python
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
POSTGRES_HOST = os.getenv("POSTGRES_HOST", "localhost")  # service name from docker-compose.yml
POSTGRES_USER = os.getenv("POSTGRES_USER")
POSTGRES_PASSWORD = os.getenv("POSTGRES_PASSWORD")
POSTGRES_DB = os.getenv("POSTGRES_DB")
POSTGRES_PORT = os.getenv("POSTGRES_PORT", "5432")
CSV_DATA_PATH = os.getenv("CSV_DATA_PATH", "/data/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv")  # default mount location
```

```python
DB_URL = f"postgresql+psycopg2://{POSTGRES_USER}:{POSTGRES_PASSWORD}@localhost:{POSTGRES_PORT}/{POSTGRES_DB}"
```

```python
try:
    engine = create_engine(DB_URL)
    print("✅ Database connection successful.")
except Exception as e:
    print("❌ Database connection failed:", e)
    exit(1)
```

```python
try:
    csv_file_path = os.path.join(CSV_DATA_PATH)  # inside /data mount
    df = pd.read_csv(csv_file_path)
    print(f"📄 Loaded CSV successfully with {len(df)} rows and {len(df.columns)} columns.")
except Exception as e:
    print("❌ Error reading CSV file:", e)
    exit(1)
```

```python
table_name = "network_data"

try:
    df.to_sql(table_name, engine, if_exists="replace", index=False)
    print(f"✅ Data successfully loaded into PostgreSQL table '{table_name}'.")
except Exception as e:
    print("❌ Error loading data to PostgreSQL:", e)
    exit(1)
```
